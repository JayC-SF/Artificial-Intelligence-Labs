```python
import numpy as np
```

```python
y1 = 1
y2 = 2
y3 = -1
x1 = np.array([1,2])
x2 = np.array([2,1])
x3 = np.array([-1,-2])
v0 = np.array([0,0])
w0 = np.array([0,0])
eta = 0.1
beta = 0.8
lam = 0

def yi_hat(w, xi):
    wxi = w@xi
    print("w @ xi:", wxi)
    return np.tanh(wxi)
def grad(w, xi, yi):
    yi_h = yi_hat(w, xi)
    print("y_hat:", yi_h)
    return 2*(yi_h-yi)*(1-yi_h**2)*xi+2*lam*w
def compute_v(v, w, xi, yi):
    gr = grad(w, xi, yi)
    print("grad:", gr)
    return beta*v+ gr

def compute_w(v_next, w):
    return w - eta*v_next
def compute_w_no_momentum(w, xi, yi):
    gr = grad(w, xi, yi)
    print("grad:", gr)
    return w - eta*gr
```

```python
v1 = compute_v(v0, w0, x1, y1)
w1 = compute_w(v1, w0)
print(v1, w1)
```
